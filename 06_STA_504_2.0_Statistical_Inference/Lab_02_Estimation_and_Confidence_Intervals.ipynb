## Detailed theory · Sampling distributions and confidence intervals

<a id="theory-01"></a>

### Normal-sample reference distributions
For independent $X_i\sim N(\mu,\sigma^2)$, $\bar X\sim N(\mu,\sigma^2/n)$ and $(n-1)S^2/\sigma^2\sim\chi^2_{n-1}$. The sample mean and sample variance are independent under this normal model. Consequently $(\bar X-\mu)/(S/\sqrt n)\sim t_{n-1}$.

<a id="theory-02"></a>

### Pivotal quantities and inversion
A pivot has a known distribution not depending on the unknown parameter, although its expression includes that parameter. Put the pivot between appropriate quantiles with probability 1−alpha, then solve the inequalities for the parameter. Record whether a quantile is a lower-tail probability or an upper-tail cutoff before substitution.

<a id="theory-03"></a>

### One-population mean
For known sigma use a z interval; for unknown sigma and independent normal observations use
$$\bar x\pm t_{1-\alpha/2,n-1}\frac{s}{\sqrt n}.$$
A large-sample approximation can be useful beyond normality under suitable moment/dependence conditions, but normality is not proved by a nonsignificant diagnostic test.

<a id="theory-04"></a>

### Difference of independent means
For unknown unequal variances, Welch's SE is $\sqrt{s_1^2/n_1+s_2^2/n_2}$ with approximate df
$$\nu=\frac{(s_1^2/n_1+s_2^2/n_2)^2}
{(s_1^2/n_1)^2/(n_1-1)+(s_2^2/n_2)^2/(n_2-1)}.$$
If a common variance is justified, pool it:
$$s_p^2=\frac{(n_1-1)s_1^2+(n_2-1)s_2^2}{n_1+n_2-2},$$
then use $s_p\sqrt{1/n_1+1/n_2}$ and n1+n2−2 df. Do not choose the pooled model merely because a low-power variance test failed to reject.

<a id="theory-05"></a>

### Paired means
Compute one difference per independent pair and use $\bar d\pm t_{1-\alpha/2,n-1}s_d/\sqrt n$. The relevant variation is the difference SD, not two independent-group SDs. Define the subtraction order. Repeated measurements on one unit are not additional independent pairs.

<a id="theory-06"></a>

### Proportions
The ordinary approximate interval uses $\hat p\pm z\sqrt{\hat p(1-\hat p)/n}$, but small or boundary counts need a better-calibrated interval. For independent-group differences, use an interval method suited to two binomial proportions rather than borrowing the pooled null SE from a hypothesis test.

<a id="theory-07"></a>

### Variance and standard deviation
In a normal sample, invert the chi-square pivot to obtain the variance interval. The larger quantile divides the numerator in the lower endpoint. Square-root endpoints for an SD interval; do not apply the same untransformed numerical bounds to SD.

<a id="theory-08"></a>

### Meaning and width
A 95% confidence procedure covers a fixed parameter in 95% of repeated samples under its assumptions. It does not mean 95% of data values lie between the bounds. Higher confidence widens an interval; more independent information usually narrows it. Interpretation is tied to the population parameter and the design.

## Worked calculation · Confidence interval: identify the pivot and show substitutions

### Question and assumptions

An independent normal sample has n=10, mean 12, sample SD 3. Construct a 95% interval for the population mean with unknown SD.

### Procedure

1. Use the t pivot with n−1=9 degrees of freedom.
2. Use t(.975,9)≈2.262157.
3. Calculate SE, margin and both bounds; state the target parameter.

### Working tables

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


n=10; xb=12; s=3; critical=2.2621571628540993; se=s/sqrt(n); margin=critical*se
show_table(["Step","Formula / substitution","Value"],[["df","n−1",n-1],["SE","3/sqrt(10)",se],["Critical t","t(.975,9)",critical],["Margin","critical × SE",margin],["Lower","12 − margin",xb-margin],["Upper","12 + margin",xb+margin]])
assert xb-margin<xb<xb+margin
```

### Decision and interpretation

The critical value is stated explicitly so the hand calculation can be checked against a t table. This interval is for the population mean; it is not a prediction interval for a new observation.
