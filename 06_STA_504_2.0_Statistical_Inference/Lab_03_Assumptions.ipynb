## Detailed theory · Hypothesis tests, errors, power and robustness

<a id="theory-01"></a>

### State a population claim
H0 and H1 concern a population/model parameter, not the observed sample mean. Equality supplies the reference boundary for many tests; directional alternatives can be written against a composite null. The alternative's direction is chosen before seeing results.

<a id="theory-02"></a>

### The testing sequence
Define the question and design; choose alpha; state assumptions; select the statistic and reference law; specify the rejection region; compute the statistic and p-value; decide; then interpret with effect magnitude and uncertainty. A critical-value calculation and p-value calculation agree only if model, tails, corrections and significance level match.

<a id="theory-03"></a>

### One-sample mean and proportion
For known sigma use $Z=(\bar X-\mu_0)/(\sigma/\sqrt n)$. For unknown sigma under a normal-sample model use $T=(\bar X-\mu_0)/(S/\sqrt n)$ with n−1 df. For a proportion, the normal null statistic is
$$Z=(\hat p-p_0)/\sqrt{p_0(1-p_0)/n}.$$
Use exact binomial inference when appropriate. Test-SE and interval-SE conventions are not automatically interchangeable.

<a id="theory-04"></a>

### Two means and paired data
For independent groups subtract the hypothesized difference and divide by Welch's or the justified pooled SE. For paired data test the mean difference using one-sample t on independent pair differences. Define direction before calculating. Treating pairs as independent loses the covariance information and changes the reference analysis.

<a id="theory-05"></a>

### Critical regions and p-values
For a two-sided t test reject for $|T|>t_{1-\alpha/2,\nu}$; for a greater alternative use the upper tail and for a less alternative the lower tail. Include observed outcomes in discrete tail sums. A p-value is not P(H0 is true) or the probability that results were caused by chance.

<a id="theory-06"></a>

### Type I error, Type II error and power
Type I error rejects a true null; alpha bounds its rate under the model. Type II error fails to reject at a specified alternative; power=1−beta. Power depends on effect size, sample size, noise, design, alpha and tails. There is no universal power value without specifying an alternative.

<a id="theory-07"></a>

### Duality with intervals
A matching two-sided level-alpha test rejects a parameter value when it lies outside the corresponding 1−alpha confidence set, subject to compatible test/interval constructions. A one-sided test needs a corresponding one-sided bound; a mismatched interval can give an apparent contradiction.

<a id="theory-08"></a>

### Diagnostics and robust choices
Independence comes primarily from design. Inspect distribution/residual plots for nonnormality, unequal variance or unusual observations. Robust or nonparametric methods must answer the intended mean, median, location or distribution question; changing a method can change the estimand.

<a id="theory-09"></a>

### Conclusions and practical importance
Use reject or do not reject rather than treating a nonsignificant result as proof of equality. A tiny p-value can accompany a negligible effect, and a large p-value can accompany an important but imprecisely estimated effect. Report effect estimates, intervals, assumptions and any multiple-testing adjustments.

## Worked calculation · Hypothesis-test decisions and power: decision table

### Question and assumptions

Test a known-SD normal mean with $H_0:\mu=0$, $H_1:\mu\ne0$, $\sigma=2$, n=25 and $\alpha=.05$. Calculate power at the specified alternative $\mu=1$.

### Procedure

1. Fix the rejection region under the null.
2. Translate the z rejection region into lower and upper sample-mean cutoffs.
3. Under mu=1, calculate probability of landing in either rejection tail.

### Working tables

These tables are calculated from the original teaching data below. They are visible before running Python and on GitHub.

| Quantity | Calculation | Value |
| --- | --- | --- |
| SE | 2/sqrt(25) | 0.4 |
| Lower rejection cutoff | −z(.975) × SE | -0.783986 |
| Upper rejection cutoff | z(.975) × SE | 0.783986 |
| Left rejection probability under mu=1 | P(mean < lower) | 4.09867e-06 |
| Right rejection probability under mu=1 | P(mean > upper) | 0.705414 |
| Power at mu=1 | Left + right | 0.705418 |
| Type II error at mu=1 | 1−power | 0.294582 |

| Truth | Reject H0 | Do not reject H0 |
| --- | --- | --- |
| H0 true | Type I error (alpha) | Correct non-rejection |
| Specified alternative true | Correct rejection (power) | Type II error (beta) |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


sigma=2; n=25; alpha=.05; se=sigma/sqrt(n); zc=NormalDist().inv_cdf(1-alpha/2)
lo=-zc*se; hi=zc*se; alternative=1
left=NormalDist(alternative,se).cdf(lo); right=1-NormalDist(alternative,se).cdf(hi)
show_table(["Quantity","Calculation","Value"],[["SE","2/sqrt(25)",se],["Lower rejection cutoff","−z(.975) × SE",lo],["Upper rejection cutoff","z(.975) × SE",hi],["Left rejection probability under mu=1","P(mean < lower)",left],["Right rejection probability under mu=1","P(mean > upper)",right],["Power at mu=1","Left + right",left+right],["Type II error at mu=1","1−power",1-left-right]])
show_table(["Truth","Reject H0","Do not reject H0"],[["H0 true","Type I error (alpha)","Correct non-rejection"],["Specified alternative true","Correct rejection (power)","Type II error (beta)"]])
assert 0<left+right<1
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng=np.random.default_rng(42)
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
ns=[10,20,30,50,80,120]; power=[]
for n in ns:
 a=rng.normal(0,8,(1200,n)); b=rng.normal(3,8,(1200,n)); power.append(np.mean(stats.ttest_ind(a,b,axis=1,equal_var=False).pvalue<.05))
plt.plot(ns,power,marker='o'); plt.axhline(.8,color='red',ls='--'); plt.xlabel('n per group'); plt.ylabel('simulated power'); plt.show()
a=rng.normal(size=60); b=np.r_[rng.normal(size=58),[8,-9]]
print('mean/median ordinary',a.mean().round(2),np.median(a).round(2),'with extremes',b.mean().round(2),np.median(b).round(2))
```
