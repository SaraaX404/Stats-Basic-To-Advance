## Detailed theory · Likelihood, moments, least squares and estimator quality

<a id="theory-01"></a>

### Likelihood is a function of the parameter
For independent observations, $L(\theta;x)=\prod f(x_i;\theta)$ treats the observed x as fixed and theta as variable. It is not a probability distribution over theta. Taking logs turns products into sums and preserves maxima where likelihood is positive.

<a id="theory-02"></a>

### Maximum-likelihood workflow
Specify support and parameter domain; write log likelihood; solve score equations for interior candidates; check curvature, boundaries and identifiability; compare candidates. Differentiating while ignoring parameter-dependent support can give a wrong answer, as with uniform-endpoint problems.

<a id="theory-03"></a>

### Poisson and exponential examples
For Poisson counts, the score is $\sum x_i/\lambda-n$ and $\hat\lambda=\bar x$, with an all-zero boundary case. For an exponential rate model $f(x;\lambda)=\lambda e^{-\lambda x}$, x≥0, the score is $n/\lambda-\sum x_i$, giving $\hat\lambda=n/\sum x_i$ for a positive total. If using a mean/scale parameter instead of a rate, its estimate is $\bar x$; declare parameterization.

<a id="theory-04"></a>

### Normal likelihood
For independent normal data, $\hat\mu_{ML}=\bar x$ and $\hat\sigma^2_{ML}=n^{-1}\sum(x_i-\bar x)^2$. The latter has expectation $(n-1)\sigma^2/n$. Replacing n by n−1 gives the usual unbiased sample variance, which is not the same finite-sample ML estimator.

<a id="theory-05"></a>

### Method of moments
Equate theoretical raw moments with $n^{-1}\sum X_i^k$ for enough k to identify the parameters. For a Poisson model the first moment gives the same estimate as ML. In an exponential-rate model the first moment $1/\lambda$ gives the reciprocal sample mean. For other models the moment estimate can differ from ML, fail to be unique or leave the parameter domain; check the solution.

<a id="theory-06"></a>

### Least squares
Specify a mean function $m_i(\beta)$ and minimize $\sum[y_i-m_i(\beta)]^2$. For an intercept-only model the minimizer is the sample mean. For a full-rank linear regression the normal equations yield OLS. Under independent common-variance normal errors, coefficient ML and OLS agree; without that model their optimization interpretations still differ.

<a id="theory-07"></a>

### Bias, variance and MSE
An unbiased rule need not be most precise. Use $MSE=Var+Bias^2$ for squared-error risk, and compare under a stated model and n. Consistency is asymptotic and does not guarantee small-sample reliability; ML is not universally unbiased or consistent without regularity.

<a id="theory-08"></a>

### Simulation checks
Generate repeated samples under known parameters, apply the estimator to each, and compare empirical expectation, bias, variance and MSE. Do not confuse variation of the original measurements with variation of repeated estimates. Increase repetitions to assess Monte Carlo uncertainty; the simulation explores the assumed model rather than proving it is appropriate.

## Theory
An estimator is a function of the random sample; its realized value is an estimate. Bias, variance, MSE and consistency describe different features.
$$MSE(\hat\theta)=Var(\hat\theta)+[E(\hat\theta)-\theta]^2.$$
For a Poisson sample, $\log L(\lambda)=\sum x_i\log\lambda-n\lambda-\sum\log(x_i!)$. Differentiating yields $\hat\lambda=\bar x$, with the all-zero boundary case included. Equating $E[X]=\lambda$ to the sample mean gives the same method-of-moments estimator.

For independent normal observations, maximum likelihood gives $\hat\mu=\bar x$ and $\hat\sigma^2_{ML}=n^{-1}\sum(x_i-\bar x)^2$. Dividing by $n-1$ produces the unbiased sample variance, rather than the ML variance.

## Worked reasoning
For a normal sample of size 5, $E[\hat\sigma^2_{ML}]=(4/5)\sigma^2$. It is biased downward at fixed sample size, while the bias tends to zero as sample size grows. Consistency and unbiasedness are distinct properties.

### Decision and interpretation

The two methods agree for this model. If all counts were zero, the maximum would occur at the boundary lambda=0, where the displayed positive-lambda derivative formula cannot be used directly.

### Your calculation check

```python
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(504)
n=5; repetitions=20000; sigma2=9
x=rng.normal(4,np.sqrt(sigma2),size=(repetitions,n))
ml=x.var(axis=1,ddof=0); unbiased=x.var(axis=1,ddof=1)
for name,a in [('ML variance',ml),('Unbiased variance',unbiased)]:
    print(name,'mean',a.mean(),'bias',a.mean()-sigma2,'MSE',np.mean((a-sigma2)**2))
counts=rng.poisson(3,size=40); lam=counts.mean()
grid=np.linspace(.05,7,400)
loglik=counts.sum()*np.log(grid)-len(counts)*grid
plt.plot(grid,loglik-loglik.max()); plt.axvline(lam,color='red',label='Sample mean / MLE')
plt.xlabel('Poisson rate'); plt.ylabel('Relative log likelihood'); plt.legend(); plt.show()
```
