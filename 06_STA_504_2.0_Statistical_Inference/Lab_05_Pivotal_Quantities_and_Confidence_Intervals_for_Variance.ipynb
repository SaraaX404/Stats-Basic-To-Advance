## Detailed theory · Pivotal inference for variance

<a id="theory-01"></a>

### Why a pivot helps
A pivotal quantity involves data and an unknown parameter but has a distribution independent of that parameter. It converts a known probability statement into an interval after algebraic inversion. Sampling distribution, pivot and observed statistic play different roles.

<a id="theory-02"></a>

### The normal-sample variance pivot
For independent normal observations,
$$Q=\frac{(n-1)S^2}{\sigma^2}\sim\chi^2_\nu,\qquad \nu=n-1.$$
The sum of squared deviations loses one degree of freedom because the sample mean was estimated. Normality is essential for this exact chi-square statement.

<a id="theory-03"></a>

### Inverting an equal-tail statement
Write
$$P\left(\chi^2_{\alpha/2,\nu}\le
\frac{\nu S^2}{\sigma^2}\le\chi^2_{1-\alpha/2,\nu}\right)=1-\alpha.$$
With lower-tail quantiles, solving gives
$$\left[\frac{\nu s^2}{\chi^2_{1-\alpha/2,\nu}},
\frac{\nu s^2}{\chi^2_{\alpha/2,\nu}}\right].$$
Taking reciprocals reverses inequality directions. A formula using upper-tail notation may look different while referring to the same quantiles.

<a id="theory-04"></a>

### Standard-deviation intervals and asymmetry
Since square root is monotone for positive values, square-root both variance endpoints to obtain the SD interval. Chi-square intervals are usually asymmetric about the estimate; forcing symmetry by writing estimate ± a single margin changes the procedure.

<a id="theory-05"></a>

### One-sided bounds and tests
Use a single quantile for a one-sided upper or lower confidence bound. To test $H_0:\sigma^2=\sigma_0^2$, evaluate $\nu s^2/\sigma_0^2$ and use the chi-square tail or tails matching the alternative. A two-sided equal-tail test checks both extremes; a large variance alternative uses the upper tail.

<a id="theory-06"></a>

### Comparing normal variances
For independent normal samples, $(S_1^2/\sigma_1^2)/(S_2^2/\sigma_2^2)$ has an F reference. Under equality the observed ratio is S1²/S2² with df n1−1,n2−1. Reversing numerator and denominator reverses the df order and tail. This test is sensitive to nonnormality; do not use it as an automatic pretest that guarantees the pooled t model.

<a id="theory-07"></a>

### Interpretation and sensitivity
Distinguish precision about the population variance from the spread of individual observations. A large n does not make the exact chi-square variance pivot valid for every skewed or heavy-tailed distribution. If the model is doubtful, choose a justified robust or resampling analysis and disclose its target and calibration.

## Theory
A pivotal quantity combines data and an unknown parameter but has a distribution that does not depend on that parameter. For an independent normal sample:
$$Q=\frac{(n-1)S^2}{\sigma^2}\sim\chi^2_{n-1}.$$
Place $Q$ between two chi-square quantiles with probability $1-\alpha$, then solve for $\sigma^2$. Inverting positive bounds reverses their order:
$$CI_{\sigma^2}=\left[\frac{(n-1)s^2}{\chi^2_{1-\alpha/2,n-1}},
\frac{(n-1)s^2}{\chi^2_{\alpha/2,n-1}}\right].$$
Square-root the endpoints for an interval for $\sigma$. This exact interval depends strongly on normality; large n alone does not make the chi-square pivot exact for arbitrary distributions.

## Interpretation
The confidence level describes how often the procedure covers the fixed parameter over repeated samples. It is not the fraction of observations inside the interval, and a realized frequentist interval is not a probability distribution on the parameter.

### Question and assumptions

For an independent normal sample, n=20 and sample variance s²=16. Construct a 95% variance interval using chi-square df=19 quantiles 8.906516 and 32.852327.

### Procedure

1. Write Q=(n−1)S²/sigma² and its chi-square reference.
2. Place Q between the .025 and .975 quantiles.
3. Solve for sigma²: the larger quantile belongs in the lower endpoint denominator.
4. Square-root the variance endpoints for an SD interval.

### Decision and interpretation

Normality is required for the exact chi-square pivot. The endpoint order reverses because solving for the variance divides by the positive chi-square bounds.
