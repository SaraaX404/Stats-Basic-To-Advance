## Detailed theory · SRS estimation, error bounds and sample size

<a id="theory-01"></a>

### Finite-population notation
For N population values let $\bar Y=N^{-1}\sum Y_i$, total $\tau=N\bar Y$ and $S^2=(N-1)^{-1}\sum(Y_i-\bar Y)^2$. Under SRS without replacement of n units, $\bar y$ estimates $\bar Y$ and $N\bar y$ estimates the total.

<a id="theory-02"></a>

### Unbiasedness and precision
Across the random sampling design, $E_d(\bar y)=\bar Y$ and
$$Var_d(\bar y)=\left(1-\frac nN\right)\frac{S^2}{n},\quad
\widehat{Var}_d(\bar y)=\left(1-\frac nN\right)\frac{s^2}{n}.$$
The finite-population correction reduces variance as the sampled fraction grows and gives zero sampling variance for a complete observed population. For the total multiply the mean variance by $N^2$. SE is the square root of an estimated variance.

<a id="theory-03"></a>

### Population proportions
Use binary values to estimate P by $\hat p=x/n$. Since the sample variance of binary observations is $n\hat p(1-\hat p)/(n-1)$,
$$\widehat{Var}_d(\hat p)=\left(1-\frac nN\right)\frac{\hat p(1-\hat p)}{n-1},\quad n>1.$$
Common large-sample versions replace n−1 by n; distinguish approximations from exact unbiased design-variance estimates.

<a id="theory-04"></a>

### Error bounds
A bound $B\approx z\,SE$ expresses a desired half-width under a justified reference approximation. A course rule such as 2SE is an approximate confidence bound, not a deterministic maximum error. Bias from coverage or response is not included in a design SE.

<a id="theory-05"></a>

### Sample size for a mean
Solve $B^2=z^2S^2(1/n-1/N)$ using the finite-population S² convention:
$$n=\frac{Nz^2S^2}{NB^2+z^2S^2}.$$
Round up and verify feasibility. If the variance is specified with denominator N instead, an algebraically different finite-N expression follows; reconcile variance definitions before comparing formulas. With a negligible sampling fraction, $n\approx z^2S^2/B^2$.

<a id="theory-06"></a>

### Sample size for a proportion and total
For a planned population proportion P, $S^2=N P(1-P)/(N-1)$ for binary values. Substituting gives
$$n=\frac{Nz^2P(1-P)}{(N-1)B^2+z^2P(1-P)}.$$
Use reliable prior/pilot information; P=.5 is conservative for a fixed absolute proportion bound. To plan a total, convert its desired bound to a mean bound by dividing by N. If N is unknown, a mean can still be estimated under an appropriate design, but a finite-population total requires population-size information or its own estimation model.

<a id="theory-07"></a>

### Practical planning
Allow for expected nonresponse, design effects and cost, and distinguish selected sample size from completed responses. In a complex design, borrowing an SRS sample-size formula without adjustment can understate the needed information. Report inputs and assumptions, not just the rounded n.

## Worked calculation · SRS estimation: total, finite-population correction and SE

### Question and assumptions

A population has N=100 units; an SRS without replacement observes values 8,10,12,14,16. Estimate the population mean and total.

### Procedure

1. Calculate the sample mean and unbiased sample variance.
2. Apply 1−n/N to the variance of the sample mean.
3. Multiply the mean by N for the total, and its variance by N².

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


x=[8,10,12,14,16]; n=len(x); N=100; xb=mean(x); ss=sum((v-xb)**2 for v in x); s2=ss/(n-1)
show_table(["Observation","Deviation","Squared deviation"],[[v,v-xb,(v-xb)**2] for v in x])
vmean=(1-n/N)*s2/n; total=N*xb
show_table(["Step","Value"],[["n",n],["N",N],["Mean estimate",xb],["Sample variance",s2],["FPC variance factor",1-n/N],["Variance of estimated mean",vmean],["SE of estimated mean",sqrt(vmean)],["Total estimate",total],["SE of total estimate",N*sqrt(vmean)]])
assert s2==10 and abs(vmean-1.9)<1e-12
```

### Decision and interpretation

These SEs describe design sampling variation, not errors caused by nonresponse or poor measurement. A normal-approximation bound such as 2SE must be labelled an approximation.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng=np.random.default_rng(42)
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
pop=rng.lognormal(4.4,.45,8000); N=len(pop); n=150; s=rng.choice(pop,n,False)
se=np.sqrt((1-n/N)*s.var(ddof=1)/n); print('estimate',s.mean().round(2),'SE',se.round(2),'approx 2-SE bound',(2*se).round(2))
z=stats.norm.ppf(.975); S=18; E=2.5; n0=z*z*S*S/E**2; nf=n0/(1+(n0-1)/N)
print('n without/with FPC:',np.ceil(n0),np.ceil(nf))
Nh=np.array([5000,3000,2000]); ph=np.array([.42,.55,.71]); W=Nh/Nh.sum(); Sh=np.sqrt(ph*(1-ph))
print('weighted p',np.sum(W*ph).round(3),'Neyman allocation',np.round(240*W*Sh/np.sum(W*Sh),1))
```
