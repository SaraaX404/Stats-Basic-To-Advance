## Detailed theory · Random variables and distributions

<a id="theory-01"></a>

### Mapping from outcomes
A random variable is a numerical function $X:S\to\mathbb R$. Outcomes remain in the underlying sample space; X maps several possible outcomes to the same number. A random vector maps to a tuple. The probability law of X follows from the probabilities of its inverse-image outcome sets.

<a id="theory-02"></a>

### Indicators and category codes
An indicator $I_A$ is 1 if A occurs and 0 otherwise, with $E[I_A]=P(A)$. Arbitrary numeric codes for colours or missingness do not create meaningful numerical distances or an ordinal scale. Separate a data-storage convention from a scientifically meaningful random variable.

<a id="theory-03"></a>

### Support and types
A discrete variable has countable possible values with probability masses. A continuous-density model assigns interval probabilities by integration, with zero mass at each individual point. Some models have both atoms and a continuous part; not every distribution is purely discrete or has a density.

<a id="theory-04"></a>

### Distribution functions
For any real-valued X, $F_X(x)=P(X\le x)$. It is nondecreasing, right-continuous and tends to 0 and 1 at the extremes. A jump at x equals $P(X=x)$. For discrete variables, strict and non-strict inequalities may differ.

<a id="theory-05"></a>

### Transformations and vectors
If $Y=g(X)$, calculate probabilities by identifying which X values map to the requested Y event. For discrete X, add masses that map to the same result. For a pair (X,Y), joint and marginal distributions describe different questions; knowing both marginals does not determine dependence.

<a id="theory-06"></a>

### Expected values
Expectation averages numerical outcomes using their probability law when the required sums or integrals exist. $E[aX+b]=aE[X]+b$, but generally $E[g(X)]\ne g(E[X])$. Decide whether the numerical coding supports the interpretation before reporting a mean.

## Worked calculation · Random variables: map outcomes to values

### Question and assumptions

Toss two independent fair coins. Let X be the number of heads.

### Procedure

1. List the four equally likely outcomes.
2. Map each outcome to X.
3. Group outcomes that produce the same value of X.

### Working tables

These tables are calculated from the original teaching data below. They are visible before running Python and on GitHub.

| Outcome | X = heads | Outcome probability |
| --- | --- | --- |
| HH | 2 | 0.25 |
| HT | 1 | 0.25 |
| TH | 1 | 0.25 |
| TT | 0 | 0.25 |

| x | Outcomes with this x | P(X=x) |
| --- | --- | --- |
| 0 | 1 | 0.25 |
| 1 | 2 | 0.5 |
| 2 | 1 | 0.25 |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


outcomes=["HH","HT","TH","TT"]
show_table(["Outcome","X = heads","Outcome probability"],[[o,o.count("H"),.25] for o in outcomes])
rows=[[k,sum(o.count("H")==k for o in outcomes),sum(o.count("H")==k for o in outcomes)/4] for k in range(3)]
show_table(["x","Outcomes with this x","P(X=x)"],rows)
assert sum(r[1] for r in rows)==4
```
