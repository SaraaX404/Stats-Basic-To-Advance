## Detailed theory · Classical and empirical probability

<a id="theory-01"></a>

### Classical probability
If N elementary outcomes are equally likely and an event contains m, its probability is m/N. The equal-likelihood assumption is part of the model. Two-dice sums are not equally likely even when the 36 ordered die pairs are.

<a id="theory-02"></a>

### Counting sequences
The multiplication counting principle multiplies the number of choices across stages. For n independent binary outcomes there are $2^n$ sequences; exactly k successes occur in $\binom nk=n!/[k!(n-k)!]$ arrangements. Ordered selections differ from combinations; replacement changes the allowed choices. A counting formula is not itself a probability until a probability model is attached.

<a id="theory-03"></a>

### Empirical probability
With k successes in n comparable trials, relative frequency is k/n. Under independent repeated trials with a stable success probability, the law of large numbers explains convergence of relative frequency to that probability. It does not guarantee that every larger prefix is closer, or that a long failure streak makes the next success more likely.

<a id="theory-04"></a>

### Precision and simulation
For Bernoulli trials, $\operatorname{Var}(\hat p)=p(1-p)/n$. A simulation estimate has Monte Carlo uncertainty; a fixed seed makes its numerical result reproducible, not exact. Use repetitions and an estimated Monte Carlo SE to distinguish a model property from simulation noise.

<a id="theory-05"></a>

### Geometric probability
For a point uniformly distributed over a region, the probability of a target is its area divided by the full area. Uniformity must be in area, not an arbitrary coordinate parameterization. For example, a radius sampled uniformly does not generate a uniformly distributed disk point.

<a id="theory-06"></a>

### Complements and interpretation
For n independent trials with success probability p, $P(\text{at least one})=1-(1-p)^n$. Independence and constant p are required. A classical probability and an empirical estimate can differ without either calculation being wrong: one is a model value and the other a random estimate.

## Worked calculation · Classical and empirical probability: convergence table

### Question and assumptions

Compare the classical probability of an even die result with relative frequencies from progressively larger reproducible simulations.

### Procedure

1. State the model probability: three even outcomes among six equally likely outcomes.
2. Count successes and divide by the number of trials at each sample size.
3. Observe that convergence is not necessarily monotonic.

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


import random
random_source=random.Random(502); outcomes=[random_source.randint(1,6) for _ in range(1000)]
rows=[]
for n in [10,50,100,500,1000]:
    k=sum(x%2==0 for x in outcomes[:n]); rows.append([n,k,k/n,.5,k/n-.5])
show_table(["Trials","Even results","Empirical probability","Model probability","Difference"],rows)
assert all(0<=r[2]<=1 for r in rows)
```

### Decision and interpretation

The model assumes a fair die. Agreement in one simulation is not proof that a physical die is fair; the data-generating process matters.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from itertools import product
rng=np.random.default_rng(42)
```

```python
def estimate_at_least_one(p,n,B=20000):
    trials=rng.random((B,n)) < p
    return np.mean(np.any(trials,axis=1))
for n in [100,1000,10000]:
    print('n=',n,'empirical P(at least one six in 4 rolls)=',estimate_at_least_one(1/6,4,n))
print('exact:',1-(5/6)**4)
print('one head in five tosses, exact:',1-(1/2)**5)
```
