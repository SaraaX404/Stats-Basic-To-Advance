## Detailed theory · Tree diagrams and sequential experiments

<a id="theory-01"></a>

### Tree structure
Each node represents information accumulated so far. Branches from that node must be exhaustive and mutually exclusive and their conditional probabilities sum to one. A leaf is a complete path, not a separate newly assumed experiment.

<a id="theory-02"></a>

### Multiplication and addition
Multiply conditional probabilities along a path to obtain its joint probability. Add disjoint leaf probabilities for an event that can occur through several paths. If two descriptions overlap, direct addition double-counts them; list complete paths first.

<a id="theory-03"></a>

### With and without replacement
With independent replacement draws, population composition resets and probabilities can remain fixed. Without replacement, both counts and denominator update. From R red and B blue balls, $P(R_1B_2)=R/(R+B)\times B/(R+B-1)$. The reversed path has a different first branch but may have the same joint probability in this symmetric drawing model.

<a id="theory-04"></a>

### Marginal and conditional questions
For second-red probability, add every path ending in red. For second-red given first-blue, work at the first-blue node instead. A marginal probability averages over earlier outcomes; a conditional probability uses the known branch.

<a id="theory-05"></a>

### Checking a tree
Check branch sums at every node and total leaf probability one. Confirm denominators correspond to the remaining objects. A simulation should reproduce the same replacement rule; independent draws with unchanged probabilities simulate a different model from sampling without replacement.

<a id="theory-06"></a>

### Repeated trials
An event such as exactly one success can be a union of several ordered sequences. When paths are equiprobable, counting may simplify the tree; otherwise add their individual products. Compact formulas should be derived from the paths rather than used without checking assumptions.

## Worked calculation · Probability tree: path multiplication and leaf table

### Question and assumptions

A bag contains three red and two blue balls. Draw two without replacement.

### Procedure

1. List the first-draw probabilities.
2. Update the second-draw counts for each first-draw result.
3. Multiply along paths and add disjoint paths for combined events.

### Working tables

These tables are calculated from the original teaching data below. They are visible before running Python and on GitHub.

| Path | First probability | Second conditional probability | Path probability |
| --- | --- | --- | --- |
| R→R | 0.6 | 0.5 | 0.3 |
| R→B | 0.6 | 0.5 | 0.3 |
| B→R | 0.4 | 0.75 | 0.3 |
| B→B | 0.4 | 0.25 | 0.1 |

| Event | Probability |
| --- | --- |
| Exactly one red | 0.6 |
| Both same colour | 0.4 |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


paths=[["R→R",3/5,2/4],["R→B",3/5,2/4],["B→R",2/5,3/4],["B→B",2/5,1/4]]
rows=[[name,p1,p2,p1*p2] for name,p1,p2 in paths]
show_table(["Path","First probability","Second conditional probability","Path probability"],rows)
show_table(["Event","Probability"],[["Exactly one red",rows[1][3]+rows[2][3]],["Both same colour",rows[0][3]+rows[3][3]]])
assert abs(sum(r[3] for r in rows)-1)<1e-12
```

### Decision and interpretation

The second branch depends on the first because no replacement occurs. Using 3/5 or 2/5 for every second-draw branch would be incorrect.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from itertools import product
rng=np.random.default_rng(42)
```

```python
bag=['R','R','B']; B=100000
without=[rng.choice(bag,2,replace=False) for _ in range(B)]
with_rep=[rng.choice(bag,2,replace=True) for _ in range(B)]
print('without replacement P(BR):',np.mean([z[0]=='B' and z[1]=='R' for z in without]))
print('exact BR:',(1/3)*1)
print('with replacement P(RR):',np.mean([z[0]=='R' and z[1]=='R' for z in with_rep]))
print('exact RR:',(2/3)**2)
print('without replacement P(at least one red):',1-(1/3)*(0/2))
```
