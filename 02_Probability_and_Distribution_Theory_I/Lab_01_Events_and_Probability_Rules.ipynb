## Detailed theory · Events and probability rules

<a id="theory-01"></a>

### Experiment, outcome and sample space
A random experiment has an uncertain individual outcome but a defined set of possibilities. An outcome $\omega$ belongs to a sample space S; an event is a subset of S. A simple event contains one outcome, a compound event several, and the empty event none. Define whether order matters before listing outcomes.

<a id="theory-02"></a>

### Set operations
$A\cap B$ means both events occur; $A\cup B$ means at least one occurs, including both. $A^c$ is everything in S outside A. The difference $A\setminus B=A\cap B^c$ means A without B. De Morgan's laws give $(A\cup B)^c=A^c\cap B^c$ and $(A\cap B)^c=A^c\cup B^c$.

<a id="theory-03"></a>

### Probability axioms and consequences
Probabilities are nonnegative, $P(S)=1$, and disjoint event probabilities add. Therefore $P(\varnothing)=0$, $P(A^c)=1-P(A)$ and
$$P(A\cup B)=P(A)+P(B)-P(A\cap B).$$
The overlap is subtracted because it is counted twice.

<a id="theory-04"></a>

### Exclusive, exhaustive and independent
Mutually exclusive events have empty intersection. Exhaustive events have union S. Independent events satisfy $P(A\cap B)=P(A)P(B)$. Two positive-probability mutually exclusive events cannot be independent: learning one occurred rules the other out. Neither exclusivity nor independence should be inferred from event names.

<a id="theory-05"></a>

### Counting and checking
For equally likely elementary outcomes, probability is favourable count divided by total count. Without equal likelihood, count outcomes with their actual probabilities. Check bounds, complement sums and whether an inclusive OR has been used. A Venn diagram illustrates membership; circle area is not automatically a calibrated probability.

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


S=set(range(1,7)); A={2,4,6}; B={4,5,6}
show_table(["Outcome","In A?","In B?","In intersection?","In union?"],[[x,x in A,x in B,x in A&B,x in A|B] for x in sorted(S)])
show_table(["Event","Probability"],[["A",len(A)/6],["B",len(B)/6],["A intersection B",len(A&B)/6],["A union B",len(A|B)/6]])
assert len(A|B)==len(A)+len(B)-len(A&B)
```

### Decision and interpretation

A and B overlap: adding their probabilities without subtracting the intersection double-counts outcomes 4 and 6.
