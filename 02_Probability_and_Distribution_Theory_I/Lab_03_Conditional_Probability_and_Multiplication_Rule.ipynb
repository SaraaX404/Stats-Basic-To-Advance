## Detailed theory · Conditional probability and multiplication

<a id="theory-01"></a>

### Conditioning changes the reference set
For $P(B)>0$,
$$P(A\mid B)=\frac{P(A\cap B)}{P(B)}.$$
B is the information supplied, so it determines the denominator. Event intersection is symmetric, but conditioning is not: generally $P(A\mid B)\ne P(B\mid A)$. The notation does not by itself specify temporal order or causation.

<a id="theory-02"></a>

### Frequency-table interpretation
In a two-way count table, a within-row percentage estimates a probability conditional on the row category. A within-column percentage conditions on the column. An overall cell proportion estimates a joint probability. The same numerator can therefore lead to different meaningful answers.

<a id="theory-03"></a>

### Multiplication rule
Rearranging the definition gives
$$P(A\cap B)=P(A\mid B)P(B)=P(B\mid A)P(A).$$
For three stages, multiply a first-stage probability by later conditional probabilities given all earlier relevant information. Omitting conditioning can assert independence without justification.

<a id="theory-04"></a>

### Independence check
If positive-probability conditioning events are independent, $P(A\mid B)=P(A)$. Equal conditional probabilities across a partition can suggest independence under the specified model; a noisy observed table is evidence, not proof. Pairwise independence of several events does not always imply their mutual independence.

<a id="theory-05"></a>

### Zero probabilities and continuous data
The elementary ratio is undefined when P(B)=0. Conditioning on an exact continuous measurement requires a conditional density or a more general construction, not division of two zero probabilities.

<a id="theory-06"></a>

### Common mistakes
Do not divide by the full population when the question says given a subgroup; do not confuse overlap with independence; and do not reverse a test's sensitivity into a posterior probability. The Bayes notebook handles that reversal explicitly.

## Worked calculation · Conditional probability: count table and changed sample space

### Question and assumptions

Of 100 students, 30 study both Mathematics and Computing, 20 Mathematics only, 10 Computing only, and 40 neither.

### Procedure

1. Build the 2×2 table with totals.
2. For P(Math|Computing), restrict the denominator to Computing students.
3. Recover the joint probability by multiplying the conditional probability by P(Computing).

### Working tables

These tables are calculated from the original teaching data below. They are visible before running Python and on GitHub.

|  | Computing | Not Computing | Total |
| --- | --- | --- | --- |
| Math | 30 | 20 | 50 |
| Not Math | 10 | 40 | 50 |
| Total | 40 | 60 | 100 |

| Calculation | Value |
| --- | --- |
| P(Math and Computing) | 0.3 |
| P(Computing) | 0.4 |
| P(Math / Computing) | 0.75 |
| P(Computing / Math) | 0.6 |
| Conditional × P(Computing) | 0.3 |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


show_table(["","Computing","Not Computing","Total"],[["Math",30,20,50],["Not Math",10,40,50],["Total",40,60,100]])
joint=30/100; conditional=30/40; pc=40/100
show_table(["Calculation","Value"],[["P(Math and Computing)",joint],["P(Computing)",pc],["P(Math | Computing)",conditional],["P(Computing | Math)",30/50],["Conditional × P(Computing)",conditional*pc]])
assert abs(conditional*pc-joint)<1e-12
```

### Decision and interpretation

Reversing the conditioning event changes the denominator: P(Math|Computing)=.75, while P(Computing|Math)=.60.
