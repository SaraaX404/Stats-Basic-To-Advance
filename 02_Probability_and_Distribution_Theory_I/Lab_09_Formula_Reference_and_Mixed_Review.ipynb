## Detailed theory · Choosing and combining probability rules

<a id="theory-01"></a>

### Translate the question first
State the experiment, outcomes, events and assumptions. The words at least, exactly, given, both and either determine different operations. Decide whether order, replacement or conditional information changes the model.

<a id="theory-02"></a>

### A problem-solving map
Use counting for equally likely elementary outcomes. Use complements for an awkward at-least-one union. Use multiplication for a joint path, total probability for several disjoint sources, and Bayes to reverse conditioning. For a discrete variable sum masses; for a continuous density integrate areas.

<a id="theory-03"></a>

### Joint, marginal and conditional quantities
A joint probability concerns two events together. A marginal probability sums or integrates over unobserved alternatives. A conditional probability uses the event known to have occurred as the denominator. The three should not share a denominator by habit.

<a id="theory-04"></a>

### Independence must be stated
With independent trials, joint event probabilities multiply without additional conditioning. Without independence, the multiplication rule still holds but uses conditional probabilities. Mutually exclusive events are not independent when both have positive probability.

<a id="theory-05"></a>

### Algebraic and numerical checks
Check bounds, probabilities summing to one over a partition, consistency of complementary events, and whether support endpoints are included correctly. Reconstruct a joint event from both conditional directions as a Bayes check. A negative variance or density integral different from one signals an error.

<a id="theory-06"></a>

### Simulation as a companion
Write down the theoretical result before simulating. Use a fixed seed for reproducibility, explain Monte Carlo variation, and inspect the data-generation mechanism. Matching one numerical answer is weak evidence if the simulation samples a different experiment.

<a id="theory-07"></a>

### Exam-style explanation
Show the event definition, rule, substitution, result and interpretation. A bare decimal or a function call does not explain why the method applies. For a model extension, state which assumption changes and redo the relevant probability law instead of altering one denominator without justification.

### Working tables

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


joint_B=.4*.05; marginal=.6*.02+joint_B; posterior_B=joint_B/marginal
show_table(["Question","Rule","Calculation","Answer"],[
["B and defect","Multiply along path",".4 × .05",joint_B],
["Any defect","Total probability",".6 × .02 + .4 × .05",marginal],
["B given defect","Bayes",".02 / .032",posterior_B],
["Defect given B","Given conditional","Model input",.05]])
assert abs(posterior_B*marginal-joint_B)<1e-12
```
