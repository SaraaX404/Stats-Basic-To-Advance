## Detailed theory · Probability density functions

<a id="theory-01"></a>

### Density and area
A density satisfies $f(x)\ge0$ and $\int_{-\infty}^\infty f(x)\,dx=1$. It can exceed 1 because its height is probability per unit of x, not a probability. A support declaration is part of the function; use zero density outside it.

<a id="theory-02"></a>

### Normalization and interval probabilities
If $f(x)=kg(x)$ on an interval, solve $k\int g(x)\,dx=1$ and check nonnegativity. Then $P(a<X<b)=\int_a^bf(x)\,dx$, intersecting the integration limits with the support. A zero-width interval gives zero probability, so strict and inclusive endpoints agree in a continuous-density model.

<a id="theory-03"></a>

### CDF and differentiation
$$F(x)=\int_{-\infty}^xf(t)\,dt.$$
Write piecewise CDF expressions below, within and above the support. Where differentiable, $F'(x)=f(x)$. A CDF is a probability in [0,1]; a density is not restricted to that height.

<a id="theory-04"></a>

### Quantiles and moments
A p-quantile solves or generalizes $F(q)=p$. For an invertible continuous CDF, simulation can use $X=F^{-1}(U)$ with U uniform on (0,1).
$$E[X]=\int xf(x)\,dx,\quad E[g(X)]=\int g(x)f(x)\,dx.$$
As with sums, an integral can fail to converge even when f is a valid density.

<a id="theory-05"></a>

### Transformations
A nonlinear change of variable does not preserve density height. For a monotone differentiable transform Y=g(X), the inverse derivative adjusts density: $f_Y(y)=f_X(g^{-1}(y))|(g^{-1})'(y)|$. Probability is preserved over corresponding regions, not pointwise heights.

<a id="theory-06"></a>

### Numerical integration and simulation
Compare a symbolic integral, numerical approximation and Monte Carlo frequency with their separate error sources. A histogram approximates density through bar area. Increasing simulation repetitions reduces Monte Carlo variation but cannot correct a wrong density or integration boundary.

## Worked calculation · PDF: density area and cumulative probability table

### Question and assumptions

Let $f(x)=2x$ for $0\le x\le1$ and zero otherwise. Find interval probabilities using $F(x)=x^2$ on the support.

### Procedure

1. Check the total integral is one.
2. Evaluate F at each interval endpoint.
3. Subtract F(lower) from F(upper); density height alone is not probability.

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


intervals=[(0,.25),(.25,.5),(.5,.75),(.75,1)]
rows=[[a,b,2*a,2*b,a*a,b*b,b*b-a*a] for a,b in intervals]
show_table(["Lower a","Upper b","f(a)","f(b)","F(a)","F(b)","P(a<X<b)"],rows)
assert abs(sum(r[-1] for r in rows)-1)<1e-12
```

### Decision and interpretation

The density reaches 2, which is allowed: only the total area must equal one. Point probability P(X=.5) is zero for this continuous distribution.
