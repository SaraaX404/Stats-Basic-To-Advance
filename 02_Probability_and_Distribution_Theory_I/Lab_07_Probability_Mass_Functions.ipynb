## Detailed theory · Probability mass functions

<a id="theory-01"></a>

### Valid PMFs
A PMF is $p(x)=P(X=x)$, nonnegative with total mass one. Define its support and state that p is zero elsewhere. A proposed symbolic PMF may need a normalization constant determined by $\sum_xp(x)=1$; normalization alone is insufficient if some values are negative.

<a id="theory-02"></a>

### Events and CDFs
For an event involving values in A, add $\sum_{x\in A}p(x)$. The CDF adds masses at or below its argument. On an integer support, $P(X>k)=1-F(k)$ while $P(X\ge k)=1-F(k-1)$. Endpoint mistakes can change small-sample p-values substantially.

<a id="theory-03"></a>

### Moments and variance
$$E[X]=\sum_xxp(x),\quad E[X^2]=\sum_xx^2p(x),\quad Var(X)=E[X^2]-E[X]^2.$$
These require the relevant moments to exist. The expectation need not be a possible observed value. Variance is nonnegative and uses squared measurement units.

<a id="theory-04"></a>

### Function expectations
For a function g, $E[g(X)]=\sum_xg(x)p(x)$. This avoids deriving a transformed PMF when only its expectation is needed. If independence is established, variances of sums add; otherwise covariance terms must be included.

<a id="theory-05"></a>

### Binomial example
For n independent trials with common success probability p, $P(X=k)=\binom nkp^k(1-p)^{n-k}$, k=0,…,n. The coefficient counts sequences, while the power product supplies their probability. The distribution of the count is not the equally likely sequence space.

<a id="theory-06"></a>

### Calculation checks
Use a contribution table for p, xp and x²p, check total probability and computed variance, and use sensible numerical precision. For large probabilities or tails, stable library survival functions can avoid subtractive cancellation in 1−CDF.

## Worked calculation · PMF: expectation and variance contribution table

### Question and assumptions

A discrete variable has probabilities .2,.5,.3 at values 0,1,2.

### Procedure

1. Check nonnegative probabilities that sum to one.
2. Calculate xp(x) and x²p(x) row by row.
3. Use Var(X)=E(X²)−E(X)².

### Working tables

These tables are calculated from the original teaching data below. They are visible before running Python and on GitHub.

| x | p(x) | x p(x) | x² p(x) |
| --- | --- | --- | --- |
| 0 | 0.2 | 0 | 0 |
| 1 | 0.5 | 0.5 | 0.5 |
| 2 | 0.3 | 0.6 | 1.2 |

| Quantity | Value |
| --- | --- |
| Sum p | 1 |
| E(X) | 1.1 |
| E(X²) | 1.7 |
| Var(X) | 0.49 |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


values=[0,1,2]; probs=[.2,.5,.3]
rows=[[x,p,x*p,x*x*p] for x,p in zip(values,probs)]
show_table(["x","p(x)","x p(x)","x² p(x)"],rows)
EX=sum(r[2] for r in rows); EX2=sum(r[3] for r in rows)
show_table(["Quantity","Value"],[["Sum p",sum(probs)],["E(X)",EX],["E(X²)",EX2],["Var(X)",EX2-EX**2]])
assert abs(sum(probs)-1)<1e-12 and abs(EX2-EX**2-.49)<1e-12
```

### Decision and interpretation

Do not confuse E(X²) with E(X)². The variance is .49, not 1.7.
