## Detailed theory · Statistics, studies and data quality

<a id="theory-01"></a>

### Statistics and its limits
Statistics turns observations into summaries and uncertain conclusions. Descriptive results describe measured cases; inference extends beyond them using a sampling or probability model. Group averages do not predict every individual, and a large dataset can still give a biased answer. The quality of a conclusion depends on the question, selection, measurement and analysis.

<a id="theory-02"></a>

### Population, unit, sample and frame
Define the population by who or what, place and time. A unit is the entity measured; a sample contains selected units; a frame is the practical list or area structure used to select them. A population parameter such as $\mu$ is fixed for the target population; a sample statistic such as $\bar x$ varies across samples. A census can still have missing units and measurement error.

<a id="theory-03"></a>

### Branches of statistical reasoning
Classical inference evaluates estimators and procedures under repeated sampling. Bayesian inference combines a likelihood with a specified prior to obtain a posterior; both approaches use assumptions and data. Parametric methods posit a distribution family; nonparametric methods relax some distributional form assumptions. Univariate and multivariate concern the number of response variables, not simply the number of spreadsheet columns.

<a id="theory-04"></a>

### Observation and intervention
Observational studies record naturally occurring exposures and responses. Experiments assign treatments and can support causal comparisons when randomization, measurement and implementation justify them. Random sampling supports population representation; random assignment supports treatment comparability. Neither guarantees the other.

<a id="theory-05"></a>

### Primary and secondary data
Primary data are collected for the present study; secondary data already exist. Check definitions, units, time period, coverage, reliability and access before reusing secondary data. A convenient available variable may not measure the intended concept.

<a id="theory-06"></a>

### Variables and scales
Nominal values label categories; ordinal values add order; interval scales have meaningful equal differences; ratio scales also have a meaningful zero. Counts are quantitative discrete; measurements often are continuous. A numeric ID remains a label. Celsius temperature is interval; mass is ratio. Scale determines whether sums, ratios and particular summaries make sense.

<a id="theory-07"></a>

### Collection, entry and cleaning
Write objectives before a questionnaire. Use short, neutral questions, explicit reference periods, response options covering the intended population and a pilot. A schedule administered by an interviewer has different interaction risks from a self-completed questionnaire. Use one unit per row, consistent variable coding and a codebook. Check missingness, range, impossible combinations and duplicates; retain an audit record rather than silently changing records.

<a id="theory-08"></a>

### Ethical interpretation and statistical misuse
Leading questions, selective denominators, truncated axes, nonrepresentative samples and selective reporting can distort conclusions. Distinguish a missing response from zero and a relative percentage change from a percentage-point change. Explain the study limitations with the findings.

<a id="theory-09"></a>

## Worked calculation · From observations to a frequency table

### Question and assumptions

A fictional survey records transport modes: bus, walk, bus, car, walk, bus. Identify the unit and construct a one-way frequency table.

### Procedure

1. Define one unit as one surveyed person. Transport mode is nominal.
2. Count each category, divide by total n, and convert to a percentage.
3. Check that counts sum to n and percentages sum to 100%.

### Working tables

These tables are calculated from the original teaching data below. They are visible before running Python and on GitHub.

| Mode | Frequency | Relative frequency | Percent |
| --- | --- | --- | --- |
| bus | 3 | 0.5 | 50 |
| car | 1 | 0.166667 | 16.6667 |
| walk | 2 | 0.333333 | 33.3333 |

| Study component | Example |
| --- | --- |
| Unit | One person |
| Variable | Transport mode |
| Scale | Nominal |
| Sample size | 6 |
| Selection risk | Sampling only at one bus stop |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


values=["bus","walk","bus","car","walk","bus"]; n=len(values)
rows=[[v,values.count(v),values.count(v)/n,100*values.count(v)/n] for v in sorted(set(values))]
show_table(["Mode","Frequency","Relative frequency","Percent"],rows)
show_table(["Study component","Example"],[["Unit","One person"],["Variable","Transport mode"],["Scale","Nominal"],["Sample size",n],["Selection risk","Sampling only at one bus stop"]])
assert sum(r[1] for r in rows)==n
```
