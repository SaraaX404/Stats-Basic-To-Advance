## Detailed theory · Tables, graphs and numerical summaries

<a id="theory-01"></a>

### Describe before calculating
For categorical data count each category, show the denominator and choose frequency or percentage. For numerical data inspect minimum, maximum, shape, clusters and outliers. A bar chart separates categories; a histogram uses adjacent numerical intervals. Bin width changes apparent shape, so inspect more than one reasonable grouping.

<a id="theory-02"></a>

### Centre and appropriate scales
The mode is the most frequent category or value and can be nonunique. The median is an ordered middle position and is resistant to extreme observations. The arithmetic mean uses every magnitude:
$$\bar x=\frac{\sum x_i}{n},\qquad \bar x_w=\frac{\sum w_ix_i}{\sum w_i}.$$
For nominal data, a mean is generally meaningless; for ordinal data, order summaries may be appropriate. A weighted mean requires meaningful nonnegative weights and a positive total weight.

<a id="theory-03"></a>

### Dispersion and denominators
Range is maximum minus minimum; $IQR=Q_3-Q_1$ describes the middle half. Population variance uses N, while the usual unbiased sample variance uses n−1:
$$s^2=\frac{\sum(x_i-\bar x)^2}{n-1},\qquad s=\sqrt{s^2}.$$
The n−1 adjustment follows the constraint $\sum(x_i-\bar x)=0$ and repeated-sampling unbiasedness for independent observations. SD describes individual variability, not the standard error of a mean.

<a id="theory-04"></a>

### Computational identity
$$\sum(x_i-\bar x)^2=\sum x_i^2-\frac{(\sum x_i)^2}{n}.$$
This is useful by hand, but subtracting nearly equal large floating-point quantities can lose accuracy. Use a stable library implementation for large-offset data.

<a id="theory-05"></a>

### Quartiles, boxplots and outlier fences
A boxplot shows the median, quartiles and whiskers under a stated convention. Tukey-type potential-outlier fences are $Q_1-1.5IQR$ and $Q_3+1.5IQR$. Whiskers commonly reach the most extreme observations inside the fences, not the fences themselves. Quantile algorithms differ, especially for small n; state the convention.

<a id="theory-06"></a>

### Standardization and relative dispersion
For population quantities $z=(x-\mu)/\sigma$; using sample mean/SD gives a descriptive sample standard score. $CV=100s/\bar x$ expresses relative dispersion only when a meaningful ratio scale and a suitable nonzero positive mean make it interpretable. It is not appropriate simply because two variables have different units.

<a id="theory-07"></a>

### Asymmetry measures
Pearson's measures use $(\bar x-\mathrm{mode})/s$ or the approximation $3(\bar x-\mathrm{median})/s$. Bowley's measure is $(Q_3+Q_1-2Q_2)/(Q_3-Q_1)$. These measure different aspects of asymmetry and can fail when their denominators vanish.

<a id="theory-08"></a>

### Choosing a summary
For roughly symmetric homogeneous numerical data, mean and SD can be useful. For skewed data, median and IQR often communicate centre and spread better. Always pair summaries with a graph; a single mean cannot reveal a mixture of populations. The next grouped-data notebook derives class-based approximations separately.

## Worked calculation · Mean and sample variance: deviations table

### Question and assumptions

For values 2, 4, 4, 6, calculate the mean, sample variance and sample SD. Distinguish the sample denominator $n-1$ from a population denominator $n$.

### Procedure

1. Add the observations and calculate the mean.
2. Calculate deviations and squared deviations.
3. Check that deviations sum to zero. Divide the squared-deviation sum by n−1 and take the square root.

### Working tables

These tables are calculated from the original teaching data below. They are visible before running Python and on GitHub.

| x | x−mean | (x−mean)^2 |
| --- | --- | --- |
| 2 | -2 | 4 |
| 4 | 0 | 0 |
| 4 | 0 | 0 |
| 6 | 2 | 4 |

| Calculation | Value |
| --- | --- |
| n | 4 |
| Mean | 4 |
| Sum of squared deviations | 8 |
| Sample variance | 2.66667 |
| Sample SD | 1.63299 |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


x=[2,4,4,6]; xb=mean(x); rows=[[v,v-xb,(v-xb)**2] for v in x]
show_table(["x","x−mean","(x−mean)^2"],rows)
SS=sum(r[2] for r in rows); variance=SS/(len(x)-1)
show_table(["Calculation","Value"],[["n",len(x)],["Mean",xb],["Sum of squared deviations",SS],["Sample variance",variance],["Sample SD",sqrt(variance)]])
assert sum(r[1] for r in rows)==0 and SS==8
```
