## Detailed theory · Shape, ranks, quantile plots and inequality

<a id="theory-01"></a>

### Central moments and shape
The empirical central moment $m_k=n^{-1}\sum(x_i-\bar x)^k$ summarizes powers of deviations. Moment skewness is $g_1=m_3/m_2^{3/2}$, with sign indicating the direction of longer-tail asymmetry in many common distributions. Quartile and moment skewness are different summaries and can disagree on complex shapes.

<a id="theory-02"></a>

### Kurtosis and sample conventions
Kurtosis $b_2=m_4/m_2^2$ gives large deviations great weight; a normal distribution has value 3. Excess kurtosis subtracts 3. Mesokurtic, leptokurtic and platykurtic describe comparison with that reference, but kurtosis should not be treated as only peak height. Bias-corrected sample definitions differ from raw empirical moment ratios. Constant data make these standardized shape measures undefined.

<a id="theory-03"></a>

### Rank correlation and ties
Rank values within each variable and average positions occupied by tied observations. Spearman correlation is the ordinary correlation of those ranks. It detects monotone association, which can be nonlinear on the original scale. It does not establish causation and can be affected by restricted ranges and ties.

<a id="theory-04"></a>

### Quantiles and Q–Q plots
Quartiles split an ordered distribution into quarters; deciles and percentiles use tenths and hundredths. A Q–Q plot compares empirical quantiles with theoretical or second-sample quantiles. Approximate straightness suggests compatibility up to location and scale; systematic curvature or tail departures reveal discrepancies. A small sample can hide departures and a large sample reveals tiny ones. A graph cannot prove a distribution assumption.

<a id="theory-05"></a>

### Time-series Z charts
Standardizing each variable as $(x-\bar x)/s$ lets series with different units share a descriptive scale. Compare timing and deviations, not original magnitudes. Compute the reference mean and SD deliberately; changing the reference period changes z values. Temporal correlation and common trends still matter after standardizing.

<a id="theory-06"></a>

### Lorenz curve construction
For nonnegative amounts with positive total, sort from smallest to largest. Plot cumulative population proportion p against cumulative amount share L(p), including (0,0) and (1,1). Equality follows the diagonal. Negative amounts or a zero total require a different interpretation and can invalidate the familiar inequality picture.

<a id="theory-07"></a>

### Gini coefficient
The area between the diagonal and the Lorenz curve, relative to the full triangle, gives
$$G=1-2\int_0^1L(p)\,dp.$$
For empirical piecewise-linear curves, use trapezoid areas. State any finite-sample correction; formulas with different normalizations need not give identical results. Equal shares give G=0; greater concentration increases G under the usual setting.

## Theory
Centre and spread do not fully describe shape. Moment skewness uses signed cubed deviations; kurtosis uses fourth powers and is sensitive to extreme observations. Sample software may apply bias corrections, so record its definition.

$$m_k=\frac1n\sum_i(x_i-\bar x)^k,\quad
g_1=m_3/m_2^{3/2},\quad b_2=m_4/m_2^2.$$
The normal reference has $b_2=3$. Excess kurtosis is $b_2-3$. Spearman's coefficient measures monotone rank association: rank both variables, averaging tied ranks, then calculate Pearson correlation of those ranks.

A Lorenz curve requires nonnegative amounts with a positive total. Sort amounts, then plot cumulative amount share against cumulative population share. Equality lies on the diagonal. Numerical integration gives $G=1-2A$, where $A$ is area under the Lorenz curve.

## Worked reasoning
For amounts 10, 10, 10, 70, the last quarter holds 70% of the total. The cumulative shares are 0, .1, .2, .3, 1. The area under the piecewise-linear curve is .275 and the Gini coefficient is .45.

## Worked calculation · Moments, ranks and Lorenz shares: reveal each contribution

### Question and assumptions

Use nonnegative amounts 10,10,10,70 to calculate shape moments and a Lorenz curve.

### Procedure

1. Calculate centred powers for the second, third and fourth moments.
2. Sort amounts and accumulate population and amount shares.
3. Calculate each trapezoid area; obtain Gini=1−2×area.

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


x=[10,10,10,70]; xb=mean(x); n=len(x)
rows=[[v,v-xb,(v-xb)**2,(v-xb)**3,(v-xb)**4] for v in x]
show_table(["x","Deviation","Squared","Cubed","Fourth power"],rows)
m2=mean(r[2] for r in rows); m3=mean(r[3] for r in rows); m4=mean(r[4] for r in rows)
total=sum(x); cumulative=0; curve=[[0,0,0,0]]
for i,v in enumerate(sorted(x),1):
    cumulative+=v; curve.append([i,v,i/n,cumulative/total])
show_table(["i","Sorted amount","Population share","Amount share"],curve)
areas=[(curve[i][2]-curve[i-1][2])*(curve[i][3]+curve[i-1][3])/2 for i in range(1,len(curve))]
show_table(["Measure","Value"],[["Moment skewness",m3/m2**1.5],["Kurtosis",m4/m2**2],["Lorenz area",sum(areas)],["Gini",1-2*sum(areas)]])
assert abs(1-2*sum(areas)-.45)<1e-12
```

### Decision and interpretation

The normal-reference kurtosis is 3, while excess kurtosis subtracts 3. These empirical moments use denominator n, not a software bias-corrected sample definition. Lorenz shares require nonnegative values and a positive total.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
x=np.array([10.,10.,10.,70.]); delta=x-x.mean()
m2=np.mean(delta**2); m3=np.mean(delta**3); m4=np.mean(delta**4)
print('Moment skewness:',m3/m2**1.5,'Kurtosis:',m4/m2**2)
share=np.r_[0,np.cumsum(np.sort(x))/x.sum()]
p=np.linspace(0,1,len(x)+1)
area=np.sum(np.diff(p)*(share[:-1]+share[1:])/2)
print('Gini:',1-2*area)
y=np.array([2,3,3,9])
print('Spearman:',stats.spearmanr(x,y).statistic)
fig,ax=plt.subplots(1,2,figsize=(10,4))
ax[0].plot(p,share,'o-',label='Lorenz'); ax[0].plot([0,1],[0,1],'--',label='Equality')
ax[0].set(xlabel='Population share',ylabel='Amount share'); ax[0].legend()
stats.probplot(x,dist='norm',plot=ax[1]); plt.tight_layout(); plt.show()
```
