## Detailed theory · Grouped and weighted frequency distributions

<a id="theory-01"></a>

### Why grouping loses information
A frequency table can represent distinct values exactly or intervals approximately. In interval data, only the count in each class is known. Any midpoint mean, grouped variance or interpolated quantile introduces assumptions about within-class observations.

<a id="theory-02"></a>

### Limits, boundaries and width
Class limits describe displayed values, such as 10–19 for whole-number records. Boundaries such as 9.5 and 19.5 describe adjacent intervals on the measurement scale. Use a consistent convention such as left-closed, right-open, with the final interval handled explicitly. Width is upper minus lower boundary.

<a id="theory-03"></a>

### Frequency density and unequal widths
When class widths vary, histogram height should be $f_i/h_i$ for count-area representation, or $f_i/(nh_i)$ for unit total area. Height alone is not frequency if widths differ. A modal class is therefore identified from density heights in an unequal-width histogram.

<a id="theory-04"></a>

### Grouped mean and variance
For midpoint $m_i$ and frequency $f_i$:
$$n=\sum f_i,\quad\bar x_g=\frac{\sum f_im_i}{n},\quad
s_g^2=\frac{\sum f_i(m_i-\bar x_g)^2}{n-1}.$$
For exact discrete-value frequency tables, replace midpoints by actual values and these formulas reproduce the raw-data summaries. For interval tables they approximate them.

<a id="theory-05"></a>

### Median and other quantiles
Find the class where cumulative frequency first reaches the target position. The common grouped median interpolation is
$$\tilde x=L+h\frac{n/2-C}{f},$$
with lower boundary L, width h, preceding cumulative count C and class frequency f. A grouped p-quantile analog uses np under that convention. The formula spreads observations uniformly within the selected class; it does not recover the unknown exact order statistics.

<a id="theory-06"></a>

### Grouped mode
For equal-width classes, with modal frequency $f_m$ and neighbours $f_{m-1},f_{m+1}$:
$$mode_g=L+h\frac{f_m-f_{m-1}}{(f_m-f_{m-1})+(f_m-f_{m+1})}.$$
Use density-height differences for unequal widths. Boundary modes, missing neighbouring classes or flat peaks require explicit qualifications; do not insert invented frequencies.

<a id="theory-07"></a>

## Theory
Grouped data tell us how many observations lie in each class, but not their exact positions. A midpoint-based mean is therefore approximate. A median or mode found by interpolation also assumes a pattern within a class.

Distinguish class limits from boundaries. Whole-number classes 25–39 and 40–54 can have boundaries 24.5–39.5 and 39.5–54.5. Adjacent histogram bars meet at their boundaries. If widths differ, plot frequency divided by class width so bar area represents frequency.

$$\bar x_g=\frac{\sum f_im_i}{\sum f_i},\qquad
\tilde x_g=L+h\frac{n/2-C}{f}.$$
For the median formula, $C$ counts observations before the median class. The fraction $(n/2-C)/f$ tells us how far across that class to move.

## Worked reasoning
With boundaries 0, 10, 20, 30 and frequencies 4, 8, 3, the midpoint mean is $(4\cdot5+8\cdot15+3\cdot25)/15=14.333$. The median class is 10–20; interpolation gives $10+10(7.5-4)/8=14.375$.

## Worked calculation · Grouped data: midpoint and cumulative-frequency table

### Question and assumptions

Boundaries 0,10,20,30 have class frequencies 4,8,3. Calculate the approximate grouped mean and interpolated median.

### Procedure

1. Use each class midpoint for the approximate mean.
2. Accumulate frequencies to locate the median class.
3. Interpolate from the lower boundary using the number of observations still needed to reach n/2.

### Working tables

These tables are calculated from the original teaching data below. They are visible before running Python and on GitHub.

| Class | Width | f | Midpoint m | fm | Cumulative f | Density f/width |
| --- | --- | --- | --- | --- | --- | --- |
| 0–10 | 10 | 4 | 5 | 20 | 4 | 0.4 |
| 10–20 | 10 | 8 | 15 | 120 | 12 | 0.8 |
| 20–30 | 10 | 3 | 25 | 75 | 15 | 0.3 |

| Calculation | Value |
| --- | --- |
| Total f | 15 |
| Sum fm | 215 |
| Grouped mean | 14.3333 |
| Median class | 10–20 |
| Interpolated median | 14.375 |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


edges=[0,10,20,30]; f=[4,8,3]; cumulative=0; rows=[]
for lo,hi,count in zip(edges[:-1],edges[1:],f):
    mid=(lo+hi)/2; cumulative+=count
    rows.append([f"{lo}–{hi}",hi-lo,count,mid,count*mid,cumulative,count/(hi-lo)])
show_table(["Class","Width","f","Midpoint m","fm","Cumulative f","Density f/width"],rows)
n=sum(f); xg=sum(r[4] for r in rows)/n; med=10+10*(n/2-4)/8
show_table(["Calculation","Value"],[["Total f",n],["Sum fm",sum(r[4] for r in rows)],["Grouped mean",xg],["Median class","10–20"],["Interpolated median",med]])
assert abs(xg-14.333333333333334)<1e-12 and med==14.375
```

### Decision and interpretation

The midpoint mean and interpolated median are approximations because the original positions within classes are unknown. When class widths differ, histogram heights must be frequency densities.

### Your calculation check

```python
import numpy as np
import matplotlib.pyplot as plt
edges=np.array([0.,10.,20.,30.]); freq=np.array([4,8,3])
mid=(edges[:-1]+edges[1:])/2
n=freq.sum(); cumulative=np.cumsum(freq)
j=np.searchsorted(cumulative,n/2); previous=cumulative[j-1] if j else 0
group_mean=np.dot(mid,freq)/n
group_median=edges[j]+(edges[j+1]-edges[j])*(n/2-previous)/freq[j]
print('Grouped mean:',group_mean,'Grouped median:',group_median)
plt.bar(edges[:-1],freq/np.diff(edges),width=np.diff(edges),align='edge',edgecolor='black')
plt.xlabel('Measurement'); plt.ylabel('Frequency density'); plt.show()
```
