## Detailed theory · Nonparametric inference and binomial tests

<a id="theory-01"></a>

### Parametric and nonparametric inference
A parametric model specifies a finite-dimensional distribution family, such as normal distributions described by a mean and variance. Nonparametric procedures may use signs, ranks, empirical distributions or category counts without specifying that full family. They still have assumptions about independence, sampling, scale and the null hypothesis. Distribution-free refers to a particular null distribution, not to freedom from all assumptions.

<a id="theory-02"></a>

### When these methods are useful
Use the scientific question first: a binary probability, paired location, independent-group ordering, distributional fit or association needs different evidence. Nonparametric procedures can suit ordinal measurements, strong nonnormality or small samples with a valid exact reference. A small sample alone does not justify any particular rank test.

<a id="theory-03"></a>

### Benefits and limitations
Signs and ranks reduce the influence of extreme magnitudes, and exact procedures can control a discrete null without a large-sample approximation. They can discard useful metric information, have coarse attainable p-values, and answer a different question from a mean comparison. A rank test does not correct biased sampling or dependent observations.

<a id="theory-04"></a>

### Binary trials and the binomial model
Let $X_i=1$ for success and 0 otherwise. With a fixed number $n$ of independent trials sharing success probability $\pi$, $X=\sum X_i$ has
$$P_\pi(X=k)=\binom nk\pi^k(1-\pi)^{n-k},\quad E[X]=n\pi,\quad Var(X)=n\pi(1-\pi).$$
Success is a label defined before analysis, not necessarily a desirable event. Sampling without replacement from a small finite population generally requires a hypergeometric model instead.

<a id="theory-05"></a>

### Hypotheses and tails
For $H_0:\pi=\pi_0$, a greater alternative uses $P_{\pi_0}(X\ge x)$ and a less alternative uses $P_{\pi_0}(X\le x)$. Include the observed count. For a two-sided test, declare the convention: probability-ordering sums null outcomes whose probability is no greater than that of the observed count. Doubling the smaller tail is another convention and can differ when $\pi_0\ne.5$.

<a id="theory-06"></a>

### Exact decisions and normal approximation
An exact p-value uses the binomial probability model, not a normal curve. Discreteness can make achieved size below the nominal $\alpha$. For a justified approximation,
$$Z=\frac{X-n\pi_0}{\sqrt{n\pi_0(1-\pi_0)}}.$$
Check both expected success and failure counts, and declare any continuity correction. Approximation accuracy depends on tail and skewness; a simple count threshold is a guideline, not a guarantee.

<a id="theory-07"></a>

### Estimation and reporting
Report $\hat\pi=x/n$, the trial definition, n, null value, alternative, exact/approximate method and an uncertainty interval appropriate for a binomial proportion. A nonsignificant result does not establish the hypothesized probability. Report the denominator and sampling restrictions so the result can be interpreted.

## Worked calculation · Exact binomial test: probability and tail table

### Question and assumptions

Test $H_0:p=.5$ against $H_1:p>.5$ after observing 8 successes among 10 independent trials. Fix $\alpha=.05$.

### Procedure

1. List k=0,…,10 and calculate each binomial probability.
2. Highlight all values k≥8, including the observed result.
3. Add the highlighted probabilities to obtain the one-sided p-value and compare it with alpha.

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


n=10; observed=8; p0=.5
rows=[[k,comb(n,k),comb(n,k)*p0**k*(1-p0)**(n-k),"yes" if k>=observed else "no"] for k in range(n+1)]
show_table(["k","n choose k","P(X=k | H0)","In right tail?"],rows)
p=sum(row[2] for row in rows if row[0]>=observed)
show_table(["Quantity","Value"],[["P(X≥8)",p],["alpha",.05],["Decision","Reject" if p<=.05 else "Do not reject"]])
assert abs(sum(row[2] for row in rows)-1)<1e-12
```

```python
n,p0=20,.5; successes=13
print('Exact two-sided binomial test:',stats.binomtest(successes,n,p0,alternative='two-sided'))
# paired changes: sign-test form of the binomial test
change=np.array([-2,1,0,3,2,-1,4,1,-3,2])
nonzero=change[change!=0]; plus=np.sum(nonzero>0)
print('nonzero pairs:',len(nonzero),'positive:',plus,'sign-test p:',stats.binomtest(plus,len(nonzero),.5).pvalue)
```
