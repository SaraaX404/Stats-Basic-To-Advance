## Detailed theory · Wilcoxon signed-rank test and sign test

<a id="theory-01"></a>

### 1. What problem do these tests solve?
There are two designs. In a one-sample problem, compare measurements with a specified location $\theta_0$. In a paired problem, compare matched measurements from the same unit or a defensible matched pair. Create one difference per independent unit. Pairing is part of the design; it is not created by sorting two unrelated samples.

For one sample set $d_i=x_i-\theta_0$; for paired observations set $d_i=\mathrm{after}_i-\mathrm{before}_i-\Delta_0$. Here $\Delta_0$ is the hypothesized change, usually zero. Positive differences mean values above the null target or increases under this subtraction order. Reversing the subtraction reverses every directional interpretation.

<a id="theory-02"></a>

### 2. What does signed-rank test?
The signed-rank null is that the distribution of the differences is symmetric about zero. In a symmetric location family, its centre is also the median, so the procedure addresses a location/median claim. It is not a general test of an arbitrary asymmetric population's median, and it does not test equality of means without additional assumptions.

Ranks reduce sensitivity to the numerical size of extreme changes while retaining information about relative magnitudes. A sign test discards magnitude information. Signed-rank often has more power when its symmetry model is appropriate; that is not a universal ranking of the two methods.

<a id="theory-03"></a>

### 3. Assumptions and preliminary checks

| Requirement | Reason | What to check |
|---|---|---|
| Independent units or pairs | Supplies independent sign information | Sampling or randomization, not a diagnostic p-value |
| Meaningful differences | Absolute differences must be comparable | Interval/ratio measurements or a justified difference scale |
| Symmetric difference distribution | Supports signed-rank location inference | A plot of the differences, substantive context and sensitivity |
| Correct matching | A pair represents the intended comparison | Unit identifiers and measurement times |
| Declared zero/tie convention | Changes effective n and the null distribution | Measurement precision, exclusions and software settings |

Normality of the raw before and after measurements is not required. What matters for paired signed-rank is the distribution of differences. With very small samples, a symmetric-looking plot is weak evidence; use knowledge of the process as well.

<a id="theory-04"></a>

### 4. Hypotheses and the direction of evidence
Let $\theta$ be the symmetric location of the difference distribution:

| Research question | Null and alternative | Evidence using $W^+$ |
|---|---|---|
| Any location change | $H_0:\theta=0$, $H_1:\theta\ne0$ | Unusually small or large |
| Increase | $H_0:\theta\le0$, $H_1:\theta>0$ | Unusually large |
| Decrease | $H_0:\theta\ge0$, $H_1:\theta<0$ | Unusually small |

For a directional composite null, calibrate at the boundary $\theta=0$ in the location-shift model. Choose the alternative before looking at the data.

<a id="theory-05"></a>

### 5. Build the ranking table
Calculate $d_i$, remove zero differences under the usual Wilcox convention, and set $n$ to the remaining number. Rank $|d_i|$ from 1 to $n$. If $t$ values occupy rank positions $a,\ldots,b$, give each their average $(a+b)/2$. Restore the original sign and place the rank in the positive or negative column:
$$W^+=\sum_{d_i>0}r_i,\qquad W^-=\sum_{d_i<0}r_i,\qquad
W^++W^-=\frac{n(n+1)}2.$$
The final identity remains true with average tied ranks after zeros are excluded. It is a useful arithmetic check.

<a id="theory-06"></a>

### 6. Statistic definitions and critical-value tables
Many two-sided lecture tables define $T=\min(W^+,W^-)$. A small $T$ indicates imbalance. For a matching two-sided table, reject when $T\le c_{n,\alpha}$. Read the table heading carefully: some tables use a one-tail significance level or a different statistic.

For a greater alternative, use the upper tail of $W^+$, or equivalently the lower tail of $W^-$. For a less alternative use the lower tail of $W^+$. Using only $\min(W^+,W^-)$ loses direction: a small minimum on the wrong side does not support the proposed increase. A table conversion between one- and two-tailed levels is valid only when that table's symmetry and tail conventions justify it. No-ties tables do not automatically apply to tied data.

<a id="theory-07"></a>

### 7. Exact inference and tied ranks
Under the symmetric null, conditional on nonzero absolute differences, each of their $2^n$ sign assignments is equally likely. Assign signs to the observed ranks, calculate the statistic for every assignment, and count assignments at least as extreme as observed. This provides a transparent exact conditional calculation even when some absolute ranks are tied. Enumeration grows exponentially, so larger samples need more efficient exact algorithms, Monte Carlo permutation or an approximation.

A two-sided definition must be stated. The worked example uses $P\{\min(W^+,W^-)\le T_{\rm obs}\}$ under sign enumeration. For directional tests sum the corresponding $W^+$ tail. Do not mechanically double any available p-value without checking the tail and discreteness convention.

<a id="theory-08"></a>

### 8. Large-sample normal approximation
With no zeros and no ties,
$$E_0(W^+)=\frac{n(n+1)}4,\qquad
\operatorname{Var}_0(W^+)=\frac{n(n+1)(2n+1)}{24}.$$
For the observed average ranks, a general conditional formula is
$$E_0(W^+)=\tfrac12\sum r_i,\qquad
\operatorname{Var}_0(W^+)=\tfrac14\sum r_i^2.$$
For nonzero observations with absolute-value tie groups of sizes $t_j$, this variance is
$$\frac{n(n+1)(2n+1)}{24}-\sum_j\frac{t_j^3-t_j}{48}.$$
Standardize $W^+$ using this mean and SD. If using a continuity correction, move the observed statistic 0.5 toward its null mean before standardizing. Label the correction and approximation; sparse, highly tied data may still need exact/permutation calibration. For all-zero differences, there is no signed-rank evidence to calculate under the zero-discarding convention.

<a id="theory-09"></a>

### 9. Zero conventions and numerical precision
Wilcox removes zeros before ranking. Pratt includes zeros when assigning ranks but excludes their signed contributions; its null mean/variance must be adjusted accordingly. The split-zero convention assigns half of zero ranks to each sign. These are different procedures. Use one declared convention throughout.

Floating-point subtraction can turn a genuine tie into two slightly different values. Round differences only to the justified recording precision, rather than choosing a rounding level that improves significance.

<a id="theory-10"></a>

### 10. Why software results can differ
Different packages may display $W^+$, $W^-$, or their minimum, and may use exact, permutation or normal-approximation p-values. They may also differ in zero treatment, tie correction and continuity correction. A different displayed statistic is not necessarily a different test. Reconcile the difference definition, effective n, rank sums, tail, and p-value method before comparing results.

<a id="theory-11"></a>

### 11. One-sample binomial sign test
For a continuous distribution with hypothesized median $\theta_0$, the null gives $P(X>\theta_0)=P(X<\theta_0)=1/2$. After applying a stated rule for exact equalities, let $N_+$ and $N_-$ be sign counts and $n=N_++N_-$. Under the null, $N_+\sim\mathrm{Binomial}(n,.5)$.

For an increase, use $P(B\ge N_+)$. For a decrease, use $P(B\le N_+)$. At $p_0=.5$, a common two-sided equal-tail value is
$$p_{\rm two}=\min\{1,\;2P(B\le\min(N_+,N_-))\}.$$
Comparing this with $\alpha$ is equivalent to comparing the undoubled smaller tail with $\alpha/2$. Do not double the probability and also halve the significance threshold. Strictly, it is the positive count $N_+$ that is binomial: $\min(N_+,N_-)$ itself does not have a full Binomial$(n,.5)$ distribution.

For discrete data with many exact-median ties, deleting ties changes the conditional sign question. A median-null statement alone does not guarantee equal nonzero sign probabilities for every discrete distribution; state the model and consider a method suited to the measurement process.

<a id="theory-12"></a>

### 12. Choosing between related tests
A paired t test targets a mean change with its own assumptions. Signed-rank targets a symmetric location shift and uses rank magnitudes. A sign test targets direction/median under its sign-probability assumptions and needs no symmetry for continuous measurements. Mann–Whitney is for independent groups; it cannot replace a paired method without losing the design.

<a id="theory-13"></a>

### 13. Magnitude, conclusion and reporting
Report the number of original pairs, zeros removed, effective n, both rank sums, statistic convention, direction, significance level, p-value method and contextual conclusion. A sample median change describes magnitude but is not automatically a signed-rank estimate of the mean.

An optional location estimate is the Hodges–Lehmann pseudomedian, the median of Walsh averages $(d_i+d_j)/2$ for $i\le j$. Under symmetry it estimates the common centre. A signed-rank-biserial summary $(W^+-W^-)/(W^++W^-)$ describes signed rank imbalance; label its zero/tie convention. A confidence interval needs a compatible inversion or justified resampling method; a p-value alone does not supply one.

## Worked calculation · Wilcoxon signed-rank test: complete working table

### Question and assumptions

Nine independent participants have before and after scores. Define $d_i=\mathrm{after}_i-\mathrm{before}_i$. Test $H_0$: the distribution of paired differences is symmetric about zero, against a two-sided location shift, at $\alpha=.05$. Differences must be suitably symmetric for the location interpretation. These are original teaching data.

### Procedure

1. Write the hypotheses, direction of difference, and significance level before calculating.
2. Compute differences. Remove the zero difference under the Wilcox convention; rank the eight nonzero absolute differences.
3. Average tied ranks: the two absolute differences of 2 occupy ranks 2 and 3, so each receives 2.5.
4. Put ranks in the positive or negative column. Sum to obtain W+ and W−; use T=min(W+,W−) for this two-sided example.
5. Enumerate all 2^8 sign assignments to obtain the exact conditional two-sided p-value. Include outcomes as extreme as the observed statistic.
6. Reject only if p≤.05; report the direction, data, assumptions and conclusion. Do not interpret this as a test of means.

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


before=[52,55,61,59,48,63,57,60,54]
after= [53,57,63,55,53,69,57,68,63]
d=[a-b for a,b in zip(after,before)]
nonzero=[x for x in d if x!=0]
ranks=average_ranks([abs(x) for x in nonzero])
rank_iter=iter(ranks); rows=[]; positive=[]; negative=[]
for i,(b,a,change) in enumerate(zip(before,after,d),1):
    if change==0:
        rows.append([i,b,a,0,0,"excluded",0,0]); continue
    r=next(rank_iter); rp=r if change>0 else 0; rm=r if change<0 else 0
    positive.append(rp); negative.append(rm)
    rows.append([i,b,a,change,abs(change),r,rp,rm])
Wp=sum(positive); Wm=sum(negative); T=min(Wp,Wm); total=sum(ranks)
rows.append(["TOTAL","","","","","",Wp,Wm])
show_table(["Pair","Before","After","d","abs(d)","Rank","Positive rank","Negative rank"],rows)
assert len(nonzero)==8 and total==8*9/2 and Wp+Wm==total
null_T=[min(sum(r*s for r,s in zip(ranks,signs)),
            total-sum(r*s for r,s in zip(ranks,signs)))
        for signs in product([0,1],repeat=len(ranks))]
extreme=sum(t<=T for t in null_T); p=extreme/len(null_T)
show_table(["Decision quantity","Value"],[
    ["Nonzero n",len(nonzero)],["Zero pairs excluded",len(d)-len(nonzero)],
    ["W+",Wp],["W−",Wm],["T=min(W+,W−)",T],
    ["Equally likely sign assignments",len(null_T)],
    ["Assignments with T ≤ observed T",extreme],["Exact two-sided p",p],
    ["alpha",.05],["Decision","Reject H0" if p<=.05 else "Do not reject H0"]])
# Sign test on the SAME pairs: magnitudes and ranks are ignored.
plus=sum(x>0 for x in nonzero); minus=sum(x<0 for x in nonzero); n=len(nonzero)
sign_p=min(1.,2*sum(comb(n,k)*.5**n for k in range(min(plus,minus)+1)))
show_table(["Sign-test step","Value"],[
    ["Positive differences",plus],["Negative differences",minus],
    ["Zeros excluded",len(d)-n],["Effective n",n],
    ["X under H0","Binomial(n, 0.5)"],["Two-sided p",sign_p],
    ["Decision","Reject H0" if sign_p<=.05 else "Do not reject H0"]])
assert plus+minus==n and 0<=p<=1 and 0<=sign_p<=1
```

```python

# Connect the exact rank distribution to the theory's normal approximation.
from itertools import product
from statistics import NormalDist
from math import sqrt

ranks=[1,2.5,2.5,4,5,6,7,8]  # Eight nonzero absolute-difference ranks.
observed_Wplus=32
n=len(ranks)
null_mean=sum(ranks)/2
null_variance=sum(r*r for r in ranks)/4
variance_from_tie_formula=n*(n+1)*(2*n+1)/24-(2**3-2)/48
assignments=[sum(r*s for r,s in zip(ranks,signs))
             for signs in product([0,1],repeat=n)]
empirical_mean=sum(assignments)/len(assignments)
empirical_variance=sum((w-empirical_mean)**2 for w in assignments)/len(assignments)
assert abs(null_mean-empirical_mean)<1e-12
assert abs(null_variance-empirical_variance)<1e-12
assert abs(null_variance-variance_from_tie_formula)<1e-12
upper=sum(w>=observed_Wplus for w in assignments)/len(assignments)
lower=sum(w<=observed_Wplus for w in assignments)/len(assignments)
two_sided=min(1,2*min(upper,lower))
distance=observed_Wplus-null_mean
correction=.5 if distance>0 else -.5 if distance<0 else 0
z=(distance-correction)/sqrt(null_variance)
normal_two_sided=2*(1-NormalDist().cdf(abs(z)))
print('E(W+):',null_mean,'Var(W+) with ties:',null_variance)
print('Exact greater p:',upper,'Exact less p:',lower)
print('Exact two-sided p:',two_sided)
print('Continuity-corrected z:',z,'Approximate two-sided p:',normal_two_sided)
assert abs(two_sided-14/256)<1e-12
```

### Decision and interpretation

The working table makes the two tests comparable: signed-rank uses magnitude ranks, while the sign test uses only direction. Their p-values need not agree. Failure to reject does not prove no change. Here the exact signed-rank calculation conditions on the observed absolute ranks and enumerates every sign assignment; it remains explicit with tied absolute differences. This is not the no-ties critical-value table distribution. A lecture critical-value table can be used only when its definition of $T$, tail and tie rules match.

**Common errors:** ranking signed differences instead of absolute differences; counting zeros in $n$; assigning separate ranks to ties; using a two-sided critical value for a one-sided question; and comparing $W^+$ with a cutoff defined for $\min(W^+,W^-)$. For decimal measurements, round differences only to the justified measurement precision before deciding ties.
