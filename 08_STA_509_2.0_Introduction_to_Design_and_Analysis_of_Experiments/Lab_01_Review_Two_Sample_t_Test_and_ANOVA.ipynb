## Detailed theory · t tests and the transition to ANOVA

<a id="theory-01"></a>

### Match the design to the question
One-sample t compares a normal population mean with a fixed target. An independent-samples analysis compares two unrelated populations; a paired analysis compares linked measurements through independent pair differences. Similar sample sizes or sorted values do not make two samples paired.

<a id="theory-02"></a>

### One-sample t
For independent normal observations with unknown variance,
$$T=\frac{\bar X-\mu_0}{S/\sqrt n}\sim t_{n-1}\quad\text{under }H_0.$$
The denominator estimates variability of the mean, not variability of an individual. Select a greater, less or two-sided alternative before data inspection.

<a id="theory-03"></a>

### Independent groups: pooled and Welch
Under a common-variance model,
$$s_p^2=\frac{(n_1-1)s_1^2+(n_2-1)s_2^2}{n_1+n_2-2},\quad
T=\frac{\bar X_1-\bar X_2-\Delta_0}{s_p\sqrt{1/n_1+1/n_2}}.$$
Use n1+n2−2 df. With unequal variances use Welch's denominator $\sqrt{s_1^2/n_1+s_2^2/n_2}$ and its approximate df. Independence remains necessary in either model.

<a id="theory-04"></a>

### Paired comparison
Define $D_i=X_i-Y_i$ according to the scientific direction. Test its mean using $(\bar D-\Delta_0)/(S_D/\sqrt n)$ with n−1 df under a normal difference model. Normality of each marginal measurement is not the paired t requirement. Correlation within pairs affects the difference variance.

<a id="theory-05"></a>

### Critical values, p-values and intervals
For a two-sided t test use both tails and a t(1−alpha/2) quantile with the correct df; equivalently use the matching confidence interval. For one-sided questions use the appropriate single tail. Report reject or do not reject, not a proof that the null is true.

<a id="theory-06"></a>

### Why several t tests are insufficient
Testing all pairs among a treatments creates $a(a-1)/2$ comparisons and inflates a family of false-positive opportunities. ANOVA supplies an omnibus equality-of-means test; an appropriate comparison procedure handles follow-up questions. Repeated unadjusted t tests are not equivalent to a single ANOVA.

<a id="theory-07"></a>

### ANOVA as a variance ratio
Treatment variation measures differences among fitted group means; error variation measures within-treatment noise. Under a suitable equal-means model, their mean squares estimate a common error variance. A large treatment/error ratio contradicts equality. With two groups and the same pooled model, the ANOVA F equals the two-sided pooled t statistic squared.

<a id="theory-08"></a>

### Reporting
Give the design, difference direction, group sizes, means, SDs, SE, df, tail, p-value or critical region, interval and practical interpretation. A statistically detectable packet-weight difference need not be a practically important manufacturing difference; define the relevant tolerance separately.

## Worked calculation · Two-sample t test: group summaries and pooled calculation

### Question and assumptions

Independent normal groups A=[8,9,10] and B=[11,12,13] are assumed to have a common variance. Test equal means two-sided at 5%.

### Procedure

1. Summarize n, mean and sample variance for each group.
2. Pool variances using their degrees of freedom.
3. Calculate the mean difference, SE, t statistic and df.
4. Compare |t| with t(.975,4)≈2.776445.

### Working tables

| --- | --- | --- | --- | --- |

| B | [11, 12, 13] | 3 | 12 | 1 |

| Step | Value |
| --- | --- |

```python

from math import comb, exp, factorial, log, sqrt
from statistics import mean, median, NormalDist
from itertools import product, combinations

def average_ranks(values):
    """Assign ranks 1..n, averaging every tie group."""
    return [1 + sum(v < x for v in values) + (sum(v == x for v in values)-1)/2
            for x in values]

def show_table(headers, rows):
    def fmt(x):
        if isinstance(x, float): return f"{x:.5g}"
        return str(x).replace("|", "/")
    lines = ["| " + " | ".join(map(str, headers)) + " |",
             "| " + " | ".join("---" for _ in headers) + " |"]
    lines += ["| " + " | ".join(fmt(x) for x in row) + " |" for row in rows]
    result = "\n".join(lines)
    try:
        from IPython.display import Markdown, display
        display(Markdown(result))
    except ImportError:
        print(result)


A=[8,9,10]; B=[11,12,13]
def s2(x):return sum((v-mean(x))**2 for v in x)/(len(x)-1)
show_table(["Group","Data","n","Mean","Sample variance"],[["A",A,len(A),mean(A),s2(A)],["B",B,len(B),mean(B),s2(B)]])
df=len(A)+len(B)-2; pooled=((len(A)-1)*s2(A)+(len(B)-1)*s2(B))/df
se=sqrt(pooled*(1/len(A)+1/len(B))); difference=mean(A)-mean(B); t=difference/se; critical=2.7764451051977987
show_table(["Step","Value"],[["A−B mean difference",difference],["Pooled variance",pooled],["SE of difference",se],["t",t],["df",df],["Two-sided 5% critical |t|",critical],["Decision","Reject equal means" if abs(t)>critical else "Do not reject"]])
assert df==4 and pooled==1 and abs(t*t-13.5)<1e-12
```

### Decision and interpretation

The sign of t follows the stated A−B direction. Under the matching two-group pooled ANOVA, F=t²=13.5. Welch uses a different SE and degrees of freedom; paired observations require a test on within-pair differences.
